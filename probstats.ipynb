```python
#1
import random
def coin_toss_experiment(coin_tosses):
    tosses = [random.choice(['Head', 'Tail']) for x in range(coin_tosses)]
    head_count = tosses.count('Head')
    tail_count = tosses.count('Tail')
    prob_head = head_count / coin_tosses
    prob_tail = tail_count / coin_tosses

    print(f"Total Tosses: {coin_tosses}")
    print(f"Number of Heads: {head_count}")
    print(f"Number of Tails: {tail_count}")
    print(f"Experimental Probability of Head: {prob_head:.2f}")
    print(f"Experimental Probability of Tail: {prob_tail:.2f}")

coin_toss_experiment(100)
```

```python
#2
coin_toss_experiment(10)
coin_toss_experiment(100)
coin_toss_experiment(1000)
coin_toss_experiment(10000)
```

```python
#3
import random
from collections import Counter

total_rolls = 6000
rolls = [random.randint(1, 6) for _ in range(total_rolls)]

counts = Counter(rolls)

print(
    f"{'Outcome':<8} | {'Observed Count':<14} | {'Experimental Prob':<18} | {'Theoretical Prob':<16}"
)
print("-" * 65)

theoretical_prob = 1 / 6

for outcome in range(1, 7):
    freq = counts[outcome]
    exp_prob = freq / total_rolls
    print(
        f"{outcome:<8} | {freq:<14} | {exp_prob:<18.4f} | {theoretical_prob:<16.4f}"
    )
```

```python
#4
import random

total_rolls = 10000

rolls = [random.randint(1, 6) for _ in range(total_rolls)]

count_A = rolls.count(1)  # Getting 1
count_B = rolls.count(6)  # Getting 6
count_A_and_B = sum(
    1 for r in rolls if r == 1 and r == 6
)  # Getting both 1 AND 6
count_A_or_B = sum(
    1 for r in rolls if r == 1 or r == 6
)  # Getting 1 OR 6

P_A = count_A / total_rolls
P_B = count_B / total_rolls
P_A_and_B = count_A_and_B / total_rolls
P_A_or_B = count_A_or_B / total_rolls

# 4. Verification
RHS = P_A + P_B - P_A_and_B

print(f"Observed P(A)               : {P_A:.4f}")
print(f"Observed P(B)               : {P_B:.4f}")
print(f"Observed P(A ∩ B)           : {P_A_and_B:.4f}")
print(f"Observed P(A ∪ B)           : {P_A_or_B:.4f}")
print(f"Calculated P(A)+P(B)-P(A∩B) : {RHS:.4f}")
print(f"Verified?                   : {P_A_or_B == RHS}")
```

```python
#5
import random

total_rolls = 10000

rolls = [random.randint(1, 6) for _ in range(total_rolls)]

count_A = sum(1 for r in rolls if r in {2, 4, 6})  # Even
count_B = sum(1 for r in rolls if r in {4, 5, 6})  # > 3
count_A_and_B = sum(
    1 for r in rolls if (r in {2, 4, 6}) and (r in {4, 5, 6})
)
count_A_or_B = sum(
    1 for r in rolls if (r in {2, 4, 6}) or (r in {4, 5, 6})
)

P_A = count_A / total_rolls
P_B = count_B / total_rolls
P_A_and_B = count_A_and_B / total_rolls
P_A_or_B = count_A_or_B / total_rolls

RHS = P_A + P_B - P_A_and_B

print(f"Observed P(A)               : {P_A:.4f}  (Theoretical: 0.5000)")
print(f"Observed P(B)               : {P_B:.4f}  (Theoretical: 0.5000)")
print(f"Observed P(A ∩ B)           : {P_A_and_B:.4f}  (Theoretical: 0.3333)")
print(f"Observed P(A ∪ B)           : {P_A_or_B:.4f}  (Theoretical: 0.6667)")
print(f"Calculated P(A)+P(B)-P(A∩B) : {RHS:.4f}")
print(f"Verified?                   : {round(P_A_or_B, 4) == round(RHS, 4)}")
```

```python
#6
import random

total_trials = 10000

# Coin: 1 = Head, 0 = Tail
# Die: 1 to 6
trials = [(random.randint(0, 1), random.randint(1, 6)) for _ in range(total_trials)]

count_A = sum(1 for coin, die in trials if coin == 1)               # Head
count_B = sum(1 for coin, die in trials if die % 2 == 0)           # Even number
count_A_and_B = sum(1 for coin, die in trials if coin == 1 and die % 2 == 0) # Head AND Even

P_A = count_A / total_trials
P_B = count_B / total_trials
P_A_and_B = count_A_and_B / total_trials

RHS = P_A * P_B

print(f"Observed P(A)           : {P_A:.4f}  (Theoretical: 0.5000)")
print(f"Observed P(B)           : {P_B:.4f}  (Theoretical: 0.5000)")
print(f"Observed P(A ∩ B)       : {P_A_and_B:.4f}  (Theoretical: 0.2500)")
print(f"Calculated P(A) * P(B)  : {RHS:.4f}  (Theoretical: 0.2500)")
print(f"Verified?               : {round(P_A_and_B, 3) == round(RHS, 3)}")
```
