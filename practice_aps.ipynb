```python
import random
def coin_toss_experiment(coin_tosses):
    tosses = [random.choice(['Head', 'Tail']) for x in range(coin_tosses)]
    head_count = tosses.count('Head')
    tail_count = tosses.count('Tail')
    prob_head = head_count / coin_tosses
    prob_tail = tail_count / coin_tosses

    print(f"Total Tosses: {coin_tosses}")
    print(f"Number of Heads: {head_count}")
    print(f"Number of Tails: {tail_count}")
    print(f"Experimental Probability of Head: {prob_head:.2f}")
    print(f"Experimental Probability of Tail: {prob_tail:.2f}")

coin_toss_experiment(100)
```

```python
coin_toss_experiment(10)
coin_toss_experiment(100)
coin_toss_experiment(1000)
coin_toss_experiment(10000)
```

```python
import numpy as np

p=0.6
sample_size = 20

samples = np.random.binomial(
    n=1,p=p, size = sample_size
)

print("customer outcomes:",sample_size)
print("number of customers:",len(samples))


number_of_clicks = np.sum(samples==1)
number_of_non_clicks = np.sum(samples ==0)
print("number of clicks:", number_of_clicks)
print("number of non clicks:",number_of_non_clicks)
print("total customer:",number_of_clicks + number_of_non_clicks)
```

```python
number_of_clicks = np.sum(samples==1)
number_of_non_clicks = np.sum(samples ==0)

empirical_click = number_of_clicks/ sample_size
empirical_non_click = number_of_non_clicks/ sample_size

theoretical_click = p
theoretical_non_click = 1-p

print("Outcome: 0 (no click)")
print("theoretical PMF:",theoretical_non_click)
print("empirical PMF:",empirical_non_click)

print("\nOutcome: 1 (click)")
print("theoretical PMF:",theoretical_click)
print("empirical PMF:",empirical_click)
```

```python

import numpy as np

p = 0.1

transactions_per_hour = 50
number_of_hours = 20000
max_flagged = 4

flagged = np.random.binomial(
    n=transactions_per_hour,
    p=p,
    size=number_of_hours
)

print("Number of simulated hours:", len(flagged))
print("First 20 hours:", flagged[:20])
print("Average flagged transactions per hour:", np.mean(flagged))
print("Maximum flagged in any hour:", np.max(flagged))

hours_within_limit = np.sum(flagged <= max_flagged)
hours_exceeding_limit = np.sum(flagged > max_flagged)

print("Hours within investigation limit:", hours_within_limit)
print("Hours exceeding investigation limit:", hours_exceeding_limit)
print("Probability of exceeding limit:",
      hours_exceeding_limit / number_of_hours)
```

```python
exceeded = np.sum(flagged > 4)

within_limit = np.sum(flagged <= 4)

p_exceeded = exceeded / len(flagged)
p_within_limit = within_limit / len(flagged)

print("Hours exceeding capacity:", exceeded)
print("Hours within capacity:", within_limit)
print("Estimated P(X > 4):", p_exceeded)
print("Estimated P(X <= 4):", p_within_limit)
```

```python

from sklearn.metrics import confusion_matrix
from sklearn.metrics import accuracy_score, precision_score, recall_score

# Actual labels
y_true = [1, 0, 1, 1, 0, 0, 1, 0]

# Predicted labels
y_pred = [1, 0, 1, 0, 0, 1, 1, 0]

# Confusion matrix
cm = confusion_matrix(y_true, y_pred, labels=[0, 1])
print("Confusion Matrix:")
print(cm)

# Calculate metrics
accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred, zero_division=0)
sensitivity = recall_score(y_true, y_pred, zero_division=0)

tn, fp, fn, tp = cm.ravel()
specificity = tn / (tn + fp) if (tn + fp) > 0 else 0

print("Accuracy:", accuracy)
print("Precision:", precision)
print("Sensitivity / Recall:", sensitivity)
print("Specificity:", specificity)
```

```python

from sklearn.metrics import f1_score

f1 = f1_score(y_true, y_pred, zero_division=0)

print("F1-score:", f1)
```

```python

import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, roc_auc_score

y_true = [0, 0, 1, 1, 0, 1, 0, 1]
y_score = [0.1, 0.4, 0.8, 0.7, 0.3, 0.9, 0.6, 0.65]

fpr, tpr, thresholds = roc_curve(y_true, y_score)
auc = roc_auc_score(y_true, y_score)

plt.plot(fpr, tpr, label=f"AUC = {auc:.2f}")
plt.plot([0, 1], [0, 1], linestyle="--", label="Random classifier")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.show()

print("AUC:", auc)
```

```python

import matplotlib.pyplot as plt
from sklearn.metrics import precision_recall_curve, average_precision_score

precision_values, recall_values, thresholds = precision_recall_curve(
    y_true, y_score
)

avg_precision = average_precision_score(y_true, y_score)

plt.plot(recall_values, precision_values,
         label=f"Average Precision = {avg_precision:.2f}")
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve")
plt.legend()
plt.show()

print("Average Precision:", avg_precision)
```

```python
import numpy as np
import matplotlib.pyplot as plt

p = 0.3
n = 10000

samples = np.random.geometric(p, size=n)

print("First 10 samples:", samples[:10])
print("Experimental mean:", np.mean(samples))
print("Theoretical mean:", 1/p)
print("Experimental variance:", np.var(samples))
print("Theoretical variance:", (1-p)/(p**2))

plt.hist(samples, bins=range(1, max(samples)+2),
         density=True, edgecolor="black")
plt.xlabel("Number of trials until first success")
plt.ylabel("Probability")
plt.title("Geometric Distribution")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

N = 20       # Total population
K = 7        # Number of defective components
n = 5        # Number selected
trials = 10000

samples = np.random.hypergeometric(
    ngood=K,
    nbad=N-K,
    nsample=n,
    size=trials
)

print("First 10 samples:", samples[:10])
print("Experimental mean:", np.mean(samples))
print("Theoretical mean:", n*K/N)

theoretical_variance = (
    n * (K/N) * (1-K/N) * ((N-n)/(N-1))
)

print("Experimental variance:", np.var(samples))
print("Theoretical variance:", theoretical_variance)

values, counts = np.unique(samples, return_counts=True)
probabilities = counts / trials

plt.bar(values, probabilities, edgecolor="black")
plt.xlabel("Number of defective components")
plt.ylabel("Experimental probability")
plt.title("Hypergeometric Distribution")
plt.show()
```
